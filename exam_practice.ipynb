## types of errors
- `ZeroDivisionError`- divide by 0
- `SyntaxError` - e.g. print "hello" or while True print("hello")
- `NameError` - not defining variable before use
- `TypeError` - '2' + 2 or when trying to mutate immutable sequences. int([list])
- `ValueError` - int("hello") - argument of the correct type, but inappropriate value. int("str")
- `IndexError` - out of range (e.g. [1, 2, 3], list[4])
- `KeyError` - dict key doesnt exist
- `AttributeError` - e.g int.reverse()
- `ImportError` - moduel not found
- `OSError` - something like filenotfound.
