## Visualizing masks

Plots the masks found by cellpose.

```python
#setup directories 
raw_dir = './raw_data'
mask_dir = './cellpose_masks'
```

```python
#sort lists since need to match down the line
raw_files = sorted([f for f in os.listdir(raw_dir) if f.endswith('.tif')])

for raw_name in raw_files:
    #this will save mask as mask_Phase_Hn249_00001.tif 
    print(raw_name)
    mask_name = f"mask_{raw_name}"
    
    raw_path = os.path.join(raw_dir, raw_name)
    mask_path = os.path.join(mask_dir, mask_name)

    if not os.path.exists(mask_path):
        print(f"Skipping {raw_name}: Mask not found.")
        continue

    # 3. Load the data
    img = tifffile.imread(raw_path)
    masks = tifffile.imread(mask_path)

    # 4. Create the Plot
    fig, ax = plt.subplots(1, 2, figsize=(12, 6))

    # Left side: RGB Mask Overlay
    # (Note: Since we aren't using the .npy colors, we let Cellpose generate them)
    overlay = plot.mask_overlay(img, masks)
    ax[0].imshow(overlay)
    ax[0].set_title(f"Overlay: {raw_name}")
    ax[0].axis('off')

    # Right side: Red Outlines
    outlines = utils.outlines_list(masks)
    ax[1].imshow(img, cmap='gray')
    for o in outlines:
        ax[1].plot(o[:, 0], o[:, 1], color='r', lw=0.5)
    ax[1].set_title("Outlines")
    ax[1].axis('off')

    plt.tight_layout()
    plt.show()
```
